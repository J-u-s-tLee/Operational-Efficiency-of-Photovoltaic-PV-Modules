```python
import PreProcessedData
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Isto é usando o scikit learn, parece a melhor opção mas não há loss nem nada
# Tenho de separar os dados só em treino e teste

data = PreProcessedData.load_data("pv_module_efficiency_dataset.csv")

# Pré-processamento
X_train, X_validation, X_test, y_train, y_validation, y_test,  _, _, _ = PreProcessedData.split_data(data, "expected_efficiency", "efficiency_level", 0.3, 42)
categorical_cols = ["pv_module_type","hotspot","birddrop","soiling","junction_box"]
numerical_cols=["affected_area","temperature","irradiance","Voc","Isc"]
X_train, X_validation, X_test= PreProcessedData.preprocess_data(X_train, X_validation, X_test, categorical_cols)

X_train_val = pd.concat([X_train, X_validation])
y_train_val = pd.concat([y_train, y_validation])
```

```python
model = LinearRegression()
model.fit(X_train_val, y_train_val)
y_train_pred = model.predict(X_train_val)

print("Train MSE:", mean_squared_error(y_train_val, y_train_pred))
print("Train R²:", r2_score(y_train_val, y_train_pred))
```

```python
y_pred = model.predict(X_test)
print("Test MSE:", mean_squared_error(y_test, y_pred))
print("Test R²:", r2_score(y_test, y_pred))
```

```python
X_train = X_train_val[['affected_area', 'Isc']]

model = LinearRegression()
model.fit(X_train, y_train_val)
y_train_pred = model.predict(X_train)

print("Train MSE:", mean_squared_error(y_train_val, y_train_pred))
print("Train R²:", r2_score(y_train_val, y_train_pred))
```
